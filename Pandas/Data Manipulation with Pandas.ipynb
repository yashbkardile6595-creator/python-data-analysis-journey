# **Data Manipulation with Pandas**

```python
# Import library
import pandas as pd
```

```python
# Read dataset
df = pd.read_csv('https://github.com/YBIFoundation/Dataset/raw/main/Titanic.csv')
```

```python
df
```

## **Missing Values**

```python
# Check for missing values
df.isna().sum()
```

```python
df.dropna()
```

```python
df.dropna(subset=['age','fare'])
```

```python
# Fill missing values - fillna()
df['age'].bfill()
df['age'].ffill()
```

```python
df['age'].mean()
```

```python
df['age'].fillna(29.8)
```

```python
df['age'].fillna(df['age'].mean())
```

# **Tips dataset**

```python
# read data
df = pd.read_csv('https://github.com/YBIFoundation/Dataset/raw/main/Tips%20Payment%20Data.csv')
```

```python
df
```

```python
# random 3 samples
df.sample(3)
```

```python
# Complete summary
df.describe(include = 'all')
```

```python
# Category of days
df['Day'].value_counts().plot(kind='pie')
```

```python
# Top 3 bills as per amount
df.nlargest(3,'Total Bill')
```

```python
# Smallest 2 bill as per amount
df.nsmallest(2, 'Total Bill')
```

```python
# Sort by bill amount then tip
df.sort_values(['Total Bill','Tip'])
```

```python
# Day wise sale
df.groupby('Day')['Total Bill'].sum()
```

```python
# day wise average sale
df.groupby('Day')['Total Bill'].mean()
```

```python
df.groupby(['Day','Gender'])['Total Bill'].mean()
```

```python
df.groupby(['Day','Gender'])[['Total Bill','Tip']].mean()
```

```python
df['Total Bill'].between(10,15)
```

```python
df[df['Total Bill'].between(10,10.5)]
```

```python
df.loc[2:5,'Tip':'Time']
```

```python
df[df['Total Bill']>15]
```
